# 3. Циклы `for` и `while`

Циклы позволяют повторять действия.

## Цикл `for`

Часто используется вместе с `range()`.

````python
for i in range(5):
    print(i)
````

`range(start, stop, step)`

- `start` — начало;
- `stop` — конец, не включается;
- `step` — шаг.

````python
print("range(2, 10, 2):")
for i in range(2, 10, 2):
    print(i)
````

## Суммирование в цикле

````python
total = 0

for i in range(1, 11):
    total += i

print("Сумма =", total)
````

## Цикл `while`

Работает, пока условие истинно.

````python
n = 5

while n > 0:
    print(n)
    n -= 1

print("Старт!")
````

## `break` и `continue`

- `break` полностью завершает цикл;
- `continue` пропускает текущую итерацию.

````python
for i in range(1, 11):
    if i == 7:
        break
    print(i)
````

````python
for i in range(1, 11):
    if i % 2 == 0:
        continue
    print(i)
````

### Задачи к блоку 6

Написать функции:

1. `is_even(n)` — возвращает `True`, если число чётное.
2. `maximum(a, b, c)` — возвращает максимум из трёх чисел.
3. `factorial(n)` — вычисляет факториал.
4. `count_vowels(text)` — считает количество гласных.
5. `is_palindrome(text)` — проверяет палиндром.
6. `mean(numbers)` — вычисляет среднее списка.
7. `normalize(values)` — делит все элементы списка на максимальный.
8. `temperature_c_to_f(c)` — переводит °C в °F.
9. `distance(x1, y1, x2, y2)` — вычисляет расстояние между двумя точками.
10. `statistics(values)` — возвращает минимум, максимум и среднее.

````python
def statistics(values):
    minimum = min(values)
    maximum = max(values)
    mean_value = sum(values) / len(values)
    return minimum, maximum, mean_value

data = [4, 10, 7, 2, 9]
mn, mx, avg = statistics(data)

print("min =", mn)
print("max =", mx)
print("mean =", avg)
````

## Мини-проект: анализ результатов эксперимента

Представим, что проведено 10 измерений некоторой величины.

Нужно:

1. Создать список результатов.
2. Найти:
   - минимум;
   - максимум;
   - среднее;
   - количество значений выше среднего.
3. Сохранить результаты в CSV.
4. Прочитать CSV с помощью `pandas`.
5. Добавить столбец отклонения от среднего.
6. Построить график результата от номера измерения.
7. Построить горизонтальную линию среднего значения.

````python
results = [10.2, 10.5, 9.8, 10.1, 10.7, 10.4, 9.9, 10.3, 10.6, 10.0]

mean_value = sum(results) / len(results)

print("Минимум:", min(results))
print("Максимум:", max(results))
print("Среднее:", mean_value)
print("Выше среднего:", sum(x > mean_value for x in results))
````

````python
import pandas as pd

experiment_df = pd.DataFrame({
    "Номер": range(1, len(results) + 1),
    "Результат": results
})

experiment_df["Отклонение от среднего"] = experiment_df["Результат"] - experiment_df["Результат"].mean()
experiment_df.to_csv("experiment_results.csv", index=False, encoding="utf-8-sig")

experiment_df
````

````python
import matplotlib.pyplot as plt

mean_value = experiment_df["Результат"].mean()

plt.figure(figsize=(9, 5))
plt.plot(
    experiment_df["Номер"],
    experiment_df["Результат"],
    marker="o",
    label="Измерения"
)
plt.axhline(mean_value, linestyle="--", label=f"Среднее = {mean_value:.2f}")
plt.xlabel("Номер измерения")
plt.ylabel("Результат")
plt.title("Результаты эксперимента")
plt.grid(True)
plt.legend()
plt.show()
````
